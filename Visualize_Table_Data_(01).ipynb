```
import pandas as pd
```

```
df = pd.read_csv('https://raw.github.com/pandas-dev/pandas/master/pandas/tests/io/data/csv/iris.csv')
df
```

# Parallel Coordinates
ดาต้าไม่เยอะมากจะโชว์ได้

```
pd.plotting.parallel_coordinates(df, 'Name'); #แยกตาม name ที่แตกต่างกัน
```

```
pd.plotting.parallel_coordinates(df, 'Name', colormap= ('Spectral'))
```

มีสองลักษณะ สีมิ้นกลีบดอกกลีบเลี้ยงยาวแต่แคบ แบบเรียว และ อีกแบบลักษณะสีเขียวเล็กกว่าสีม่วง และกลมกว่า

## ปรับ parameter: color

```
pd.plotting.parallel_coordinates(df, 'Name', color=('#FF00E0', '#00FFE2', '#ECFF00')) #html color
```

```
pd.plotting.parallel_coordinates(df, 'Name', color=('c', 'm', 'r')) #ตัวย่อของสี R G B W C=ฟ้า M=ม่วง Y K=ดำ
```

###ใช้ค่าสี แบบอยากได้สีนี้กี่เปอมาบวกกับสีนี้กี่เปอ

(x,x,x,x) x ตัวสุดท้ายคือ alpha ค่าโปร่งแสงของสี

```
pd.plotting.parallel_coordinates(df, 'Name', color=((0.5,0.8,1,0.5), (0.8,0.7,1,1), (0.6,0.7,1,1))) #rgb value (r,g,b) | r,g,b -> [0,1]
```

## ปรับ parameter: colormap

```
pd.plotting.parallel_coordinates(df, 'Name', colormap='cool')
```

```
pd.plotting.parallel_coordinates(df, 'Name', colormap='tab10')
```

# เริ่ม Matplotlib

```
from matplotlib import pyplot as plt
```

### แปลง data จาก pandas dataframe เป็น numpy array

```
np_data = df.iloc[:,:-1].to_numpy()
np_data
```

# Visualize array data ด้วย pixel

```
import matplotlib
```

แปลงข้อมูลตัวเลขให้เป็นค่าสี

```
matplotlib.rcParams['figure.figsize']=[15,15] #กำหนดขนาดของ figure

plt.imshow(np_data[:50,:],interpolation='nearest')
```

```
plt.imshow(np_data[50:100,:],interpolation='nearest')
```

# การใช้ subplot

```
plt.subplot(1,3,1)  # subplot(มีกี่แถว, มีกี่หลัก, เป็นตัวที่เท่าไหร่แถวที่เท่าไหร่)
plt.imshow(np_data[:50,:],interpolation='nearest',cmap='magma')

plt.subplot(1,3,2)
plt.imshow(np_data[50:100,:],interpolation='nearest',cmap='magma')

plt.subplot(1,3,3)
plt.imshow(np_data[100:150,:],interpolation='nearest',cmap='magma')
```

```
plt.imshow(np_data[:10,:],interpolation='nearest',cmap='magma',alpha = 0.5 )
```
